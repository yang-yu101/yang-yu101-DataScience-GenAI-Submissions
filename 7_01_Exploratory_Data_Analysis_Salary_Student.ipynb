### Visualising the Data

**Task 8 — Bar chart.** Create a bar chart comparing the **average salary in USD** across experience levels (`EN`, `MI`, `SE`, `EX`). Add clear axis labels and a title. Which group has the highest average?

```python
avg_salary_in_USD = DP.groupby("experience_level")["salary_in_usd"].mean()

avg_salary_in_USD
```

```python
avg_salary_in_USD = avg_salary_in_USD.reindex(["EN", "MI", "SE", "EX"])

# Plot the aggregated averages directly
avg_salary_in_USD.plot(kind='bar', color='tab:blue')

plt.xlabel("Experience Level")
plt.ylabel("Average Salary (USD)")
plt.title("Average Salary by Experience Level")
plt.xticks(rotation=0)  # Keep labels horizontal

plt.show()
```

From the bar chart, EX got the highest average salary

**Task 9 — Scatter plot.** Make a scatter plot of `work_year` against `salary_in_usd`. Label the axes. What patterns do you notice? Are any points difficult to distinguish because they overlap?

```python
plt.scatter(DP['work_year'], DP['salary_in_usd'])
plt.xlabel('work_year')
plt.ylabel('salary_in_usd')
plt.title('work_year vs salary_in_usd')
plt.plot()
```

Salaries appear to be higher in more recent years, especially in 2023 and 2024. There are also some very high salaries that may be outliers.

Many points overlap, particularly in 2023 and 2024, making it difficult to distinguish individual data points.

**Task 10 — Distribution.** Plot the distribution of `salary_in_usd`. Does it look approximately normal, skewed, or something else? Are there extreme values?

```python

sns.displot(DP['salary_in_usd'], bins=30, height=5, aspect=1.8)

plt.xlabel("salary_in_usd")
plt.ylabel("Count")
plt.title("Distribution of Salary in USD")
plt.show()
```

right skewed distribution
